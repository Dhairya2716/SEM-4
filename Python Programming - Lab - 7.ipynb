### 09) WAP to check whether two given strings are anagram or not using set.

```python
def are_anagrams(str1, str2):
    
    str1 = str1.replace(" ", "").lower()
    str2 = str2.replace(" ", "").lower()
    
    return set(str1) == set(str2)

string1 = "listen"
string2 = "silent"

if are_anagrams(string1, string2):
    print(f"'{string1}' and '{string2}' are anagrams.")
else:
    print(f"'{string1}' and '{string2}' are not anagrams.")
```

### 11) WAP to count number of vowels in given string using set.

```python
count = 0
vowels = {'a', 'e', 'i', 'o', 'u'}  
s = input("Enter a string: ").lower()

for char in s:
    if char in vowels: 
        count += 1

print("Number of vowels:", count)
```

### 12) WAP to check if a given string is binary string or not.

```python
def check_binary_string(binary_string):
    for char in binary_string:
        if char != '0' and char != '1': 
            return f'Given string "{binary_string}" is not binary.'
    return f'Given string "{binary_string}" is binary.'

binary_string = input("Enter a binary string: ")

result = check_binary_string(binary_string)
print(result)
```

### 13) WAP to sort dictionary by key or value.

```python
s1 = {101:'DJ',105:'sd',102:'poihjv',103:'aa'}
user_input = input('want to sort according to "key" or "value" : ').lower()

def sort_dict(user_input):
    if user_input == 'key':
        return dict(sorted(s1.items(),key = lambda item: item[0]))
    elif user_input == 'value':
        return dict(sorted(s1.items(),value = lambda item: item[1].lower()))
    else:
        return 'wrong input'

sort_dict(user_input)
```

### 15) WAP to handle missing keys in dictionaries.
#### Example : Given, dict1 = {'a': 5, 'c': 8, 'e': 2} 
#### if you look for key = 'd', the message given should be 'Key Not Found', otherwise print the value of 'd' in dict1.

```python
dict2 = {'a': 5, 'c': 8, 'e': 2}
get_key = input("Enter value to find key: ")

key_found = False

for key, value in dict2.items():
    if key == get_key:
        print(f"Key: {key}, Value: {value}")
        key_found = True
        break

if not key_found:
    print("Key not found")
```
