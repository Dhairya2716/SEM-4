```python

choose = int(input("write 1 to find celcius or 2 to find farenfiet : "))


if(choose == 2):
    c = int(input("Enter value of celcius : "))
    fahrenhrit = ((c*(9/5))+32)
    print(f"{c} celcius is equal to {fahrenhrit} fahrenhrit")
    
if(choose == 1):
    f = int(input("Enter value of farenheit : "))
    celcius = (f-32)*(5/9)
    print(f"{f} fahrenhrit is equal to {celcius} celcius")
```

### 09) WAP to find the distance between two points in 2-D space.

```python
import math

x1 = int(input("Enter value 1st x point : "))
y1 = int(input("Enter value 1st y point : "))
x2 = int(input("Enter value 2nd x point : "))
y2 = int(input("Enter value 2nd y point : "))

distance = math.sqrt((x2-x1)**2-(y2-y1)**2)
print(f"distance between the points is {distance}")
```

### 10) WAP to print sum of n natural numbers.

```python
n = int(input("Enter value to print first natural number : "))
natural = n
sum = 0
while(n != 0):
    sum += n
    n -= 1

print(f"sum of first {natural} number is {sum}")
```

### 11) WAP to print sum of square of n natural numbers.

```python
n = int(input("Enter value to print first natural number : "))
natural = n
sum = 0
while(n != 0):
    sum += n**2
    n -= 1

print(f"sum of first {natural} number is {sum}")
```

### 13) WAP to swap two numbers.

### 14) WAP to get the distance from user into kilometer, and convert it into meter, feet, inches and centimeter.

```python
distance = int(input("Enter value of distance in kilometer : "))

meter = distance * 1000
centimeter = meter * 100
inch = centimeter * 2.54
feet = centimeter * 0.032

print(f"kilometer to meter is {meter}")
print(f"kilometer to centimeter is {centimeter}")
print(f"kilometer to inch is {inch}")
print(f"kilometer to feet is {feet}")
```
