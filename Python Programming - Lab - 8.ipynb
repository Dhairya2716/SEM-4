### 01) Write a function to calculate BMI given mass and height. (BMI = mass/h**2)

### 02) Write a function that add first n numbers.

### 03) Write a function that returns 1 if the given number is Prime or 0 otherwise.

```python
def prime_or_not(n):
    if n < 2:
        return 0
    else:
        for i in range(2,n):
            if n % i == 0:
                return 0
        return 1

print(prime_or_not(13))
```

### 04) Write a function that returns the list of Prime numbers between given two numbers.

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def get_prime_numbers(start, end):
    return [n for n in range(start, end + 1) if is_prime(n)]

start = 2
end = 20
print(get_prime_numbers(start, end))
```

### 05) Write a function that returns True if the given string is Palindrome or False otherwise.

```python
def palidrome_or_not(s):
    if s == s[::-1]:
        return f'given string {s} is palindrome'
    else:
        return f'given string {s} is not a palindrome'

print(palidrome_or_not('121'))
```

### 08) Write a recursive function to find nth term of Fibonacci Series.

```python
def fibonacci(a,b,n):
    if n==1:
        return a
    return fibonacci(b,a+b,n-1)

print(fibonacci(0,1,5))
```

### 10) Write a function to get the sum of the scores ending with zero.
#### Example : scores = [200, 456, 300, 100, 234, 678] 
#### Ans = 200 + 300 + 100 = 600

```python
scores = [200, 456, 300, 100, 234, 678] 

def sum_of_ending_zero_nums(scores):
    sum = 0
    for i in scores:
        if i % 10 == 0:
            sum += i
    return sum

print(sum_of_ending_zero_nums(scores))
```

### 11) Write a function to invert a given Dictionary.
#### hint: keys to values & values to keys
#### Before : {'a': 10, 'b':20, 'c':30, 'd':40}
#### After : {10:'a', 20:'b', 30:'c', 40:'d'}

```python
dict2 = {'a':10,'b':20,'c':30,'d':40}

def swap_key_and_values_in_dict(dict2):
    dict3 = {}
    for i,j in dict2.items():
        dict3[j] = i
    return dict3

print(swap_key_and_values_in_dict(dict2))
```

### 12) Write a function to check whether the given string is Pangram or not.
#### hint: Pangram is a string containing all the characters a-z atlest once.
#### "the quick brown fox jumps over the lazy dog" is a Pangram string.

```python
s1 = "the quick brown fox jumps over the lazy dog"

def pangram_or_not(s1):
    alphabets = 'abcdefghijklmnopqrstuvwxyz'
    s1 = s1.replace(' ','').lower()
    s1 = list(set(s1))
    s1.sort()
    s1 = ''.join(s1)
    if s1 == alphabets:
        return 'given String is Pangram'
    else:
        return 'given String is not a Pangram'

pangram_or_not(s1)
```

### 13) Write a function that returns the number of uppercase and lowercase letters in the given string. 
#### example : Input : s1 = AbcDEfgh ,Ouptput : no_upper = 3, no_lower = 5

```python
s1 = "AbcDEfgh"
def countUPLW(s1):
    lower=0
    upper=0
    for i in s1:
        if i.islower():
            lower+=1
        else:
            upper+=1
    print(f"Lower case are:{lower}")
    print(f"Upper case are:{upper}")

countUPLW(s1)
```

### 15) For the given list of names of students, extract the names having more that 7 characters. Use filter().

```python
  l1 = ['dhairya','malay','kalpuuuuu','meet@vas','rahul','sachin']

def has_more_than_7_chars(name):
    return len(name) > 7

def extract_names(l1):
    return list(filter(has_more_than_7_chars, l1))

print(extract_names(l1))
```

### 16) For the given list of names of students, convert the first letter of all the names into uppercase. use map().

```python
l1 = ['dhairya','malay','kalp','meet']

def cap(l1):
    return list(map(str.capitalize, l1))

print(cap(l1))
```
