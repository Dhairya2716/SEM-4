### 04) WAP to find the longest word(s) in a file

```python
longest_word = ''
with open("file-1.txt", "r") as fp:
    for line in fp:
        words = line.split() 
        for word in words:
            if len(word) > len(longest_word): 
                longest_word = word

print("Longest word: ", longest_word)
print("Length: ", len(longest_word))
```

### 10) WAP to write first 100 prime numbers to a file named primenumbers.txt 
(Note: each number should be in new line)

```python
def is_prime(num):
    if num < 2:
        return False
    for i in range(2,int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

def prime_numbers(n, filename):
    count = 0
    num = 2
    primes = []
    while count < n:
        if is_prime(num):
            primes.append(num)
            count += 1
        num += 1

    with open(filename,'w') as fp:
        for prime in primes:
            fp.write(f"{prime}\n")

print(prime_numbers(100,"primenumbers.txt"))
```

### 13) Demonstrate tell() and seek() for all the cases(seek from beginning-end-current position) taking a suitable example of your choice.

```python
def seek_tell_demo():
    with open("seekandtell.txt", "w") as f:
        f.write("Hello, this is a test file for seek and tell demonstration.")

    with open("seekandtell.txt", "rb") as f:
        print("Initial position:", f.tell())  
        
        print("Reading 5 characters:", f.read(5))
        print("Position after reading 5 characters:", f.tell())
        
        f.seek(0, 0)
        print("Position after seeking to the beginning:", f.tell())
        
        f.seek(10, 0)
        print("Position after seeking to 10th byte from beginning:", f.tell())
        
        f.seek(5, 1)
        print("Position after seeking 5 bytes forward from current position:", f.tell())
        
        f.seek(0, 2)
        print("Position after seeking to end:", f.tell())

seek_tell_demo()
```
