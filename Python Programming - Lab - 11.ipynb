### 05) WAP to generate 3 random integers between 100 and 999 which is divisible by 5.

```python
def get_three_random_number():
    count = 0
    num = []
    while count != 3:
        random_value = random.randint(100,999)

        if random_value % 5 == 0:
            num.append(random_value)
            count += 1

    return num

print(get_three_random_number())
```
