### 01) WAP to handle following exceptions:
1. ZeroDivisionError
2. ValueError
3. TypeError
#### Note: handle them using separate except blocks and also using single except block too.

```python
try:
    a = int(input("Enter no1 : "))
    b = int(input("Enter no2 : "))
    print(a/b)
except (ZeroDivisionError, ValueError, TypeError) as err:
    print(type(err))
    print(err)
```

### 07) WAP to create an udf divide(a,b) that handles ZeroDivisionError.

```python
def zerodivisonerror(a,b):
    try:
        c = a // b
        return c
    except ZeroDivisionError as err:
        errorgenerated = (type(err).__name__  +  " : " + str(err))
        return errorgenerated

num1 = int(input("Enter 1st number : "))
num2 = int(input("Enter 2nd number : "))
zerodivisonerror(num1, num2)
```

### 08) WAP that gets an age of a person form the user and raises ValueError with error message: "Enter Valid Age" :
#### If the age is less than 18. 
#### otherwise print the age.

```python
def invalidage(num):
    try:
        if num > 18:
            print(f"your age ${num} is valid")
        else:
            raise ValueError('Age must be > than 18')
    except ValueError as err:
        print(type(err).__name__ + " : " + str(err))
agenum = int(input("Enter your age : "))
invalidage(agenum)
```
