# Tuple

### 01) WAP to find sum of tuple elements.

```python
t1 = (1,2,3,4,5)
sum = 0
for i in t1:
    sum += i
print(sum)
```

### 02) WAP to find Maximum and Minimum K elements in a given tuple.

```python
t2 = (1,5,3,2,9,7,4,6,8,0)
sorted_t2 = sorted(t2)
k = 2
minimum = sorted_t2[:k]
maximum = sorted_t2[-k:]
print(tuple(minimum))
print(tuple(maximum))
```

### 03) WAP to find tuples which have all elements divisible by K from a list of tuples.

```python
l1 = (1,2,3,4,5,15,20,8,9)
result = set()
k = int(input("Enter a number greater than 0 and less than {}".format(len(l1))))
for i in l1:
    if i % k == 0:
        result.add(i)
print(result)
```

### 04) WAP to create a list of tuples from given list having number and its cube in each tuple.

```python
numbers = (1,2,3,4)
cubes = [(nums,nums ** 3) for nums in numbers]
print(cubes)
```

### 05) WAP to find tuples with all positive elements from the given list of tuples.

```python
t3 = (-1,-8,5,3,-9,8,6,2,-1)
output = []
for i in t3:
    if i > 0:
        output.append(i)
print(tuple(output))
```

### 06) WAP to add tuple to list and vice – versa.

```python
t4 = (1,2,3,4)
l2 = [5,6,7,8]
l2.append(t4)
t4 = t4 + tuple(l2)
print(l2)
print(t4)
```

### 07) WAP to remove tuples of length K.

```python
list_t = [(1,2,3,4,8,5,6),(77,89,56)]
k = 3
for i in list_t:
    if len(i) == k:
        print(i)
```

### 08) WAP to remove duplicates from tuple.

```python
t5 = (1,1,2,3,4,5,5,6,6,8,9,8,8,1)
result = set(t5)
print(tuple(result))
```

### 09) WAP to multiply adjacent elements of a tuple and print that resultant tuple.

```python
t4 = (1,2,3,4)
print(*([t4[i] * t4[i+1] for i in range(len(t4)-1)]),)
```

### 10) WAP to test if the given tuple is distinct or not.

```python
orignal_t = (1,2,3,4,5,6,7,1)
new_t = set(orignal_t)
if len(orignal_t) == len(new_t):
    print("given tuple is distinct")
else:
    print("given tuple is not distinct")
```
