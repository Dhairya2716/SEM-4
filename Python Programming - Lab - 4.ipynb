### 05) WAP to print even length word in string.

```python
s = 'Dhairya is good'
l1 = []
result = ''
l1 = s.split()
for i in l1:
    if(len(i) % 2 == 0):
        result = result + ' '+ i
print(result)
```

### 06) WAP to count numbers of vowels in given string.

```python
s = 'Dhairya is good boy'
vowels = 'aeiou'
count = 0
for char in s:
    if char.lower() in vowels:
        count = count + 1
print(count)
```

### 07) WAP to capitalize the first and last character of each word in a string.

```python
sentence = "Dhairya is good"
l1 = sentence.split()
newWord=""
for i in l1:
    newWord = newWord + " " +  i[:1].upper() + i[1:len(i)-1].lower() + i[len(i)-1:len(i)].upper()
print(newWord)
```

### 10) : Display credit card number. 
### card no. : 1234 5678 9012 3456
### display as  : **** **** **** 3456

```python
card_no = "1234 5678 9012 3456"
masked_card_no = ""


last_four_digits = card_no[-4:]


masked_count = 0

for char in card_no:
    if char == " ":
        masked_card_no += " "
    elif char.isdigit():
        if masked_count < len(card_no.replace(" ", "")) - 4:
            masked_card_no += "*"
            masked_count += 1
        else:
            masked_card_no += char
    else:
        masked_card_no += char

print(masked_card_no)
```

### 12) : Rearrange the given string. First lowercase then uppercase alphabets.
### input : EHlsarwiwhtwMV
### output : lsarwiwhtwEHMV

```python
s = input("Enter a string : ")
lower = ''
upper = ''
for i in s:
    if(i.isupper()):
        upper += i
    else:
        lower += i
print(lower + upper)
```
